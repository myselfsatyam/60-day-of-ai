# Day 8 — How Machines Read Human Language
## Text Preprocessing using Python, NLTK, and Regex

### Objective
Build a text preprocessing pipeline that converts noisy real-world text into machine-readable tokens.

```python
import nltk
import re
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

# Step 1 — Load Raw Text Samples

We use:
- Social media text
- News article text
- Customer review text

These samples contain:
- Emojis
- Punctuation
- Stopwords
- Inconsistent casing

```python
social_media = """
OMG!!! 😂😂 I can't believe this phone is sooo good!!!
Best purchase everrrr 💯🔥 #awesome #tech
"""

news_article = """
BREAKING: The Government announced new climate policies on Tuesday.
Experts say the reforms could reduce emissions by 40% before 2030.
"""

customer_review = """
I ordered this laptop last week, and honestly, it's not worth the price.
Battery life is terrible!!! 😡 The screen quality is decent though.
"""

samples = {
    "Social Media": social_media,
    "News Article": news_article,
    "Customer Review": customer_review
}
```

# Step 2 — Tokenisation

Tokenisation breaks text into:
- Sentences
- Words

This helps NLP systems process text efficiently.

```python
for name, text in samples.items():

    print(f"\n{name}")
    print("=" * 50)

    sentence_tokens = sent_tokenize(text)
    word_tokens = word_tokenize(text)

    print("\nSentence Tokens:")
    print(sentence_tokens)

    print("\nWord Tokens:")
    print(word_tokens)
```

# Step 3 — Stopword Removal

Stopwords are common words like:
- the
- is
- and

These words usually add little meaning.

```python
stop_words = set(stopwords.words('english'))

for name, text in samples.items():

    words = word_tokenize(text.lower())

    filtered_words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]

    print(f"\n{name}")
    print("=" * 50)

    print("\nBefore:")
    print(words)

    print("\nAfter:")
    print(filtered_words)

    print("\nVocabulary Before:", len(set(words)))
    print("Vocabulary After:", len(set(filtered_words)))
```

# Step 4 — Remove Punctuation using Regex

Regex helps clean noisy symbols and punctuation.

This improves consistency for NLP models.

```python
for name, text in samples.items():

    cleaned_text = re.sub(r'[^a-zA-Z\s]', '', text)

    print(f"\n{name}")
    print("=" * 50)

    print("\nBefore Cleaning:")
    print(text)

    print("\nAfter Cleaning:")
    print(cleaned_text)
```

# Step 5 — Stemming vs Lemmatization

- Stemming cuts words mechanically
- Lemmatization converts words into meaningful root forms

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

text = "The striped bats are hanging on their feet for best performances."

words = word_tokenize(text.lower())

words = [word for word in words if word.isalpha()]

print(f"{'Original':15} {'Stemmed':15} {'Lemmatized':15}")
print("-" * 45)

for word in words:

    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)

    print(f"{word:15} {stemmed:15} {lemmatized:15}")
```

# Why Text Preprocessing Matters

Text preprocessing converts noisy human language into structured data.

Without preprocessing:
- Models learn inconsistent vocabulary
- Noise increases
- Accuracy decreases
- Semantic understanding becomes weaker

Preprocessing improves:
- NLP model accuracy
- Search quality
- Sentiment analysis
- Text classification
- Chatbot understanding
